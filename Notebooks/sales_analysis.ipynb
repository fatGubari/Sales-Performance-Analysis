```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Data/raw_sales.csv", encoding='latin1')
df.head()
```

```python
# Number of rows and columns
print(df.shape)
# Column names
print(df.columns)
# Data types
print(df.info())
# Statistical summary
print(df.describe())
# Missing values
print(df.isnull().sum())
# Duplicate records
print(df.duplicated().sum())
```

```python
df.columns = (df.columns.str.strip().str.lower().str.replace(" ", "_").str.replace("-","_"))
print(df.columns)
```

```python
print(df[["order_date", "ship_date"]].dtypes)
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])
print(df[["order_date", "ship_date"]].dtypes)

# or

# date_cols = ["order_date", "ship_date"]

# for col in date_cols:
#     if col in df.columns:
#         df[col] = pd.to_datetime(df[col])
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())

if df.duplicated().sum():
    print(df[df.duplicated(keep=False)].head())

df.drop_duplicates()
```

```python
negative_sales = df[df["sales"] < 0]
print(len(negative_sales))
display(negative_sales.head())
```

```python
# Extract date components
df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["month_name"] = df["order_date"].dt.month_name()
```

```python
# Calculate profit margin
df["profit_margin"] = (df["profit"] / df["sales"] * 100)
```

```python
# Calculate shipping duration
df["shipping_date"] = (df["ship_date"] - df["order_date"]).dt.days
```

```python
# Check for invalid quantities
invalid_qty = df[df["quantity"] <= 0]
print(len(invalid_qty))
```

```python
# Check for invalid discounts
invalid_disc = df[(df["discount"]<0) | (df["discount"] > 1)]
print(len(invalid_disc))
```

```python
# Check for illogical dates
if ("order_date" in df.columns) and ("ship_date" in df.columns):
    bad_dates = df[df["ship_date"] < df["order_date"]]
    print(len(bad_dates))
```

```python
df.to_csv("../Data/cleaned_sales.csv", index=False)
```
